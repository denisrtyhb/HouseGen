```python
from pathlib import Path
from IPython.display import SVG, display, HTML

ROOT = Path("scripts/outputs")
IDXS = [0, 1, 2]  # change indices to browse

def show_row(i: int):
    gt_svg = ROOT / "gt" / f"{i}c_0_gt.svg"
    pred_svg = ROOT / "pred" / f"{i}c_0_pred.svg"
    gg = ROOT / "graphs_gt" / f"{i}.svg"
    gp = ROOT / "graphs_pred" / f"{i}.svg"

    missing = [p for p in (gt_svg, pred_svg, gg, gp) if not p.exists()]
    if missing:
        print(f"Skipping {i} (missing): {missing}")
        return

    display(HTML(f"<h3>Sample index <code>{i}</code></h3>"))
    display(HTML("<b>floorplan gt | floorplan pred</b>"))
    display(
        HTML(
            '<table><tr valign="top">'
            + f"<td>{SVG(gt_svg.read_text(encoding='utf-8'))._repr_svg_()}</td>"
            + f"<td>{SVG(pred_svg.read_text(encoding='utf-8'))._repr_svg_()}</td>"
            + "</tr></table>"
        )
    )
    display(HTML("<b>graph gt | graph pred</b>"))
    display(
        HTML(
            '<table><tr valign="top">'
            + f"<td>{SVG(gg.read_text(encoding='utf-8'))._repr_svg_()}</td>"
            + f"<td>{SVG(gp.read_text(encoding='utf-8'))._repr_svg_()}</td>"
            + "</tr></table>"
        )
    )

for i in IDXS:
    show_row(i)
```
